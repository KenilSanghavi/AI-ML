```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

df = None
last_plot = None

def load_dataset():
    global df

    print("\n== Load Dataset ==")

    file_path = input("Enter the path of the dataset (CSV file): ")

    try:
        df = pd.read_csv(file_path)
        print("Dataset loaded successfully!")
    except Exception as e:
        print("Error:", e)

def explore_data():
    global df

    if df is None:
        print("Please load a dataset first.")
        return

    while True:
        print("\n== Explore Data ==")
        print("1. Display first 5 rows")
        print("2. Display last 5 rows")
        print("3. Display column names")
        print("4. Display data types")
        print("5. Display basic info")
        print("6. Back")

        choice = input("Enter your choice: ")

        if choice == "1":
            print(df.head())

        elif choice == "2":
            print(df.tail())

        elif choice == "3":
            print(df.columns)

        elif choice == "4":
            print(df.dtypes)

        elif choice == "5":
            print(df.info())

        elif choice == "6":
            break

        else:
            print("Invalid choice.")


def dataframe_operations():
    global df

    if df is None:
        print("Please load dataset first.")
        return

    while True:
        print("\n== DataFrame Operations ==")
        print("1. Sort Data")
        print("2. Filter Data")
        print("3. Add New Column")
        print("4. Delete Column")
        print("5. Back")

        choice = input("Enter your choice: ")

        if choice == "1":

            col = input("Enter column name: ")

            try:
                print(df.sort_values(by=col))
            except:
                print("Invalid column.")

        elif choice == "2":

            col = input("Enter column name: ")
            value = input("Enter value: ")

            try:
                print(df[df[col].astype(str) == value])
            except:
                print("Invalid operation.")

        elif choice == "3":

            col_name = input("New column name: ")
            value = input("Default value: ")

            df[col_name] = value
            print("Column added successfully.")

        elif choice == "4":

            col = input("Column name to delete: ")

            if col in df.columns:
                df.drop(columns=[col], inplace=True)
                print("Column deleted.")
            else:
                print("Column not found.")

        elif choice == "5":
            break

        else:
            print("Invalid choice.")


def handle_missing_data():
    global df

    if df is None:
        print("Please load dataset first.")
        return

    while True:

        print("\n== Handle Missing Data ==")
        print("1. Display rows with missing values")
        print("2. Fill missing values with mean")
        print("3. Drop rows with missing values")
        print("4. Replace missing values with a specific value")
        print("5. Back")

        choice = input("Enter your choice: ")

        if choice == "1":

            print(df[df.isnull().any(axis=1)])

        elif choice == "2":

            numeric_cols = df.select_dtypes(include='number').columns

            for col in numeric_cols:
                df[col].fillna(df[col].mean(), inplace=True)

            print("Missing values filled with mean.")

        elif choice == "3":

            df.dropna(inplace=True)
            print("Rows with missing values removed.")

        elif choice == "4":

            value = int(input("Enter replacement value: "))

            df.fillna(value, inplace=True)
            print("Missing values replaced.")

        elif choice == "5":
            break

        else:
            print("Invalid choice.")


def descriptive_statistics():
    global df

    if df is None:
        print("Please load dataset first.")
        return

    print("\n== Descriptive Statistics ==")

    numeric_df = df.select_dtypes(include='number')

    if numeric_df.empty:
        print("No numeric columns found.")
        return

    for col in numeric_df.columns:
        print(f"\n----- {col} -----")
        print(f"Count    : {numeric_df[col].count()}")
        print(f"Mean     : {numeric_df[col].mean():.2f}")
        print(f"Median   : {numeric_df[col].median():.2f}")

        mode_val = numeric_df[col].mode()
        if not mode_val.empty:
            print(f"Mode     : {mode_val.iloc[0]}")

        print(f"Min      : {numeric_df[col].min()}")
        print(f"Max      : {numeric_df[col].max()}")
        print(f"Range    : {numeric_df[col].max() - numeric_df[col].min()}")
        print(f"Variance : {numeric_df[col].var():.2f}")
        print(f"Std Dev  : {numeric_df[col].std():.2f}")
        print(f"Q1       : {numeric_df[col].quantile(0.25):.2f}")
        print(f"Q2       : {numeric_df[col].quantile(0.50):.2f}")
        print(f"Q3       : {numeric_df[col].quantile(0.75):.2f}")

def data_visualization():
    global df, last_plot

    if df is None:
        print("Please load dataset first.")
        return

    while True:

        print("\n== Data Visualization ==")
        print("1. Bar Plot")
        print("2. Line Plot")
        print("3. Scatter Plot")
        print("4. Pie Chart")
        print("5. Histogram")
        print("6. Stack Plot")
        print("7. Back")

        choice = input("Enter your choice: ")

        if choice == "1":

            x = input("Enter x-axis column name: ")
            y = input("Enter y-axis column name: ")

            plt.figure(figsize=(8, 5))
            plt.bar(df[x], df[y])
            plt.title("Bar Plot")
            plt.xticks(rotation=45)

            last_plot = plt.gcf()
            plt.show()

            print("Bar plot displayed successfully!")

        elif choice == "2":

            x = input("Enter x-axis column name: ")
            y = input("Enter y-axis column name: ")

            plt.figure(figsize=(8, 5))
            plt.plot(df[x], df[y], marker='o')
            plt.title("Line Plot")

            last_plot = plt.gcf()
            plt.show()

            print("Line plot displayed successfully!")

        elif choice == "3":

            print("\n== Scatter Plot ==")

            x = input("Enter x-axis column name: ")
            y = input("Enter y-axis column name: ")

            print("Generating scatter plot...")

            plt.figure(figsize=(8, 5))
            plt.scatter(df[x], df[y])
            plt.xlabel(x)
            plt.ylabel(y)
            plt.title("Scatter Plot")

            last_plot = plt.gcf()
            plt.show()

            print("Scatter plot displayed successfully!")

        elif choice == "4":

            col = input("Enter category column name: ")

            pie_data = df[col].value_counts()

            plt.figure(figsize=(8, 5))
            plt.pie(pie_data.values,labels=pie_data.index,autopct="%1.1f%%")
            plt.title("Pie Chart")

            last_plot = plt.gcf()
            plt.show()

            print("Pie chart displayed successfully!")

        elif choice == "5":

            col = input("Enter numeric column name: ")

            plt.figure(figsize=(8, 5))
            plt.hist(df[col], bins=10)

            plt.title("Histogram")

            last_plot = plt.gcf()
            plt.show()

            print("Histogram displayed successfully!")

        elif choice == "6":

            numeric_cols = df.select_dtypes(include=np.number).columns

            if len(numeric_cols) < 2:
                print("Need at least two numeric columns.")
                continue

            plt.figure(figsize=(8, 5))

            plt.stackplot(range(len(df)),*[df[col] for col in numeric_cols],labels=numeric_cols)

            plt.legend(loc='upper left')
            plt.title("Stack Plot")

            last_plot = plt.gcf()
            plt.show()

            print("Stack plot displayed successfully!")

        elif choice == "7":
            break

        else:
            print("Invalid choice.")


def save_visualization():
    global last_plot

    print("\n== Save Visualization ==")

    if last_plot is None:
        print("No visualization available.")
        return

    filename = input("Enter file name (example: scatter_plot.png): ")

    last_plot.savefig(filename)

    print("Visualization saved successfully!")

while True:

    print("\n========== Data Analysis & Visualization Program ==========")
    print("1. Load Dataset")
    print("2. Explore Data")
    print("3. Perform DataFrame Operations")
    print("4. Handle Missing Data")
    print("5. Generate Descriptive Statistics")
    print("6. Data Visualization")
    print("7. Save Visualization")
    print("8. Exit")
    print("==========================================================")

    choice = input("\nEnter your choice: ")

    if choice == "1":
        load_dataset()

    elif choice == "2":
        explore_data()

    elif choice == "3":
        dataframe_operations()

    elif choice == "4":
        handle_missing_data()

    elif choice == "5":
        descriptive_statistics()

    elif choice == "6":
        data_visualization()

    elif choice == "7":
        save_visualization()

    elif choice == "8":
        print("Program terminated successfully.")
        break

    else:
        print("Invalid choice. Please try again.")
```
